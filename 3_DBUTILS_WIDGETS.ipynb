### Learning dbutils widgets

```python
dbutils.widgets.help()
```

```python
dbutils.widgets.text("country",'')
```

```python
country=dbutils.widgets.get("country")
print(country)
print(f'Retrieving {country} data  from db')
```

```python
dbutils.widgets.text("range_value","5")
```

```python
int_range=int(dbutils.widgets.get("range_value"))
print(int_range)
for i in range(int_range):
    print(i)
```

```python
dbutils.widgets.dropdown("Gender","Select",["Select","Male","Female"])
```

```python
gender=dbutils.widgets.get("Gender")
print(gender)
```

```python
dbutils.widgets.combobox("Country","India",["Select","India","USA","Germany"])
```

```python
Country=dbutils.widgets.get("Country")
print(f'Country selected is {Country}')
```
