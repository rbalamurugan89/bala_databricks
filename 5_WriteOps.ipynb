## mode - append , overwrite , errorifexists, ignore 
default mode -> errorifexists

overwrite - it will remove the exisitng data overwrite with new data 

append - adding new data , adding  a new file 

ignore - do nothing 

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("custdetails").getOrCreate()
ddl_str="custid int, fname string, lname string, age int, profession string"
cust_df=spark.read.schema(ddl_str).csv("/Volumes/izwd37dev/wd37db/rawdatta/custs")
cust_df.write.mode("ignore").csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout",header=True,sep="|")
```

```python
cust_df.write.mode("append").csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout",sep="|",header=True)
```

```python
#write as parquet format

cust_ddl="custid int,fname string,lname string,age int,profession string"
cust_df=spark.read.schema(cust_ddl).csv("dbfs:/Volumes/izwd37dev/wd37db/rawdatta/custs")
cust_df.show(4)
display(cust_df.take(5))
cust_df.write.mode("overwrite").parquet("dbfs:/Volumes/izwd37dev/wd37db/rawdatta/cust_parquet_out")
```

## ORC - Optmized Row columnar File Format 
#### simlar to Parquet

```python
#/Volumes/izwd37dev/wd37db/rawdatta/custs
#read from csv and write as orc format
cust_schema="custid int,fname string,lname string,age int,profession string"
cust_orc_df=spark.read.schema(cust_schema).csv("/Volumes/izwd37dev/wd37db/rawdatta/custs",schema=cust_schema)
cust_orc_df.show(2)

cust_orc_df.write.mode("append").orc("/Volumes/izwd37dev/wd37db/rawdatta/cust_orc_out")
```

```python
# programatically - create a dataframe 
# list[tuple]

data=[(100,"laptop",150000,10),(200,"biscuit",2000,90),(201,"cycle",2000,24),(220,"bike",10000,4)]
data_schema=["id","product","price","quantity"]

df=spark.createDataFrame(data,data_schema)
df.show(2)
df.printSchema()

#write into table

df.write.saveAsTable("izwd37dev.wd37db.tbl_products")

```
